```python
def count_words():
    with open("document.txt", "r") as file:
        return len(file.read().split())

print(count_words())
```

```python
def find_and_replace(old_word, new_word):
    with open("data.txt", "r") as file:
        text = file.read()

    updated_text = text.replace(old_word, new_word)

    with open("data.txt", "w") as file:
        file.write(updated_text)

    print(updated_text)

find_and_replace("to", "the")
```

```python
def print_lines_in_reverse():
    with open("reverse.txt", "r") as file:
        lines = file.readlines()

    for line in reversed(lines):
        print(line, end="")

print_lines_in_reverse()
```

```python
def print_file_stats():
    with open("stats.txt", "r") as file:
        text = file.read()

    print("Lines:", len(text.splitlines()))
    print("Words:", len(text.split()))
    print("Characters:", len(text))

print_file_stats()
```
